# Notebook to generate a PSTH plot.

This notebook will generate a PSTH plot, by default for Unit 100 for session
2022-06-11 of monkey P.

It needs the spike sorting NWB file for that session in
`../../cache/dandi_data/spikesorting/` and the behavior table
`../../cache/behavior_processing/ring.csv`. See the top-level README for how to
get both.

```python
"""Imports."""

import sys
from pathlib import Path

import pandas as pd
from matplotlib import pyplot as plt

import run_ring

# The repository root (where paths.py lives), found from this notebook's folder.
_root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'paths.py').exists())
sys.path.insert(0, str(_root))
import paths
from phys_processing import spike_counts as spike_counts_lib
```

```python
"""Constants."""

BEHAVIOR_CACHE_PATH_RING = paths.RING_CSV
PALETTE = {
    (0,): (1, 0, 0),
    (1,): (0, 1, 0),
    (2,): (0, 0, 1),
    (0, 1): (0.7, 0.7, 0),
    (0, 2): (0.7, 0, 0.7),
    (1, 2): (0, 0.7, 0.7),
}
DASHES = {
    0: (1.0, 0.0),
    1: (1.0, 1.0),
    2: (3.0, 1.5),
}
STIM_TIME_WINDOW = (0, 2500)  # Time window for stimulus phase in ms
CUE_TIME_WINDOW = (0, 500)  # Time window for cue phase in ms, relative to cue onset
```

```python
"""Load data."""

SUBJECT = "Perle"
SESSION = "2022-06-11"

ring_behavior = pd.read_csv(BEHAVIOR_CACHE_PATH_RING)
session_behavior = ring_behavior[
    (ring_behavior["subject"] == SUBJECT) &
    (ring_behavior["session"] == SESSION)
].reset_index(drop=True)
```

```python
"""Generate PSTH plot."""

UNIT = 126

# Load spike counts and trials
spike_counts, trials = spike_counts_lib.unit_spike_counts(SUBJECT, SESSION, UNIT)
unit_info = spike_counts_lib.units(SUBJECT, SESSION).iloc[UNIT]

# Generate figure
fig = run_ring.generate_psth(spike_counts, trials, session_behavior)
title = f"Unit {UNIT}, {unit_info.probe}/{unit_info.quality}"
fig.suptitle(title)
plt.tight_layout()
fig.subplots_adjust(wspace=0.2, hspace=0.4)
```
